## 63. ndarray subclass with a name attribute

class NamedArray(np.ndarray):
    def __new__(cls, data, name="array"):
        obj = np.asarray(data).view(cls)
        obj.name = name
        return obj
    def __array_finalize__(self, obj):

## 80. Fixed-size centered crop with fill padding

``` python
def centered_crop(A, center, shape, fill=0):
    out = np.full(shape, fill, dtype=A.dtype)
    src, dst = [], []
    for n, c, size in zip(A.shape, center, shape):
        start, end = c - size//2, c - size//2 + size
        lo, hi = max(start, 0), min(end, n)
        src.append(slice(lo, hi))
        dst.append(slice(lo-start, hi-start))
    out[tuple(dst)] = A[tuple(src)]
    return out

print(centered_crop(np.arange(16).reshape(4, 4), (0, 0), (3, 3), -1))
```

    def __setitem__(self, key, value):

## 88. Game of Life

``` python
def life_step(board):
    b = board.astype(int)
    neighbors = sum(np.roll(np.roll(b, i, 0), j, 1)
                    for i in (-1, 0, 1) for j in (-1, 0, 1)
                    if (i, j) != (0, 0))
    return ((neighbors == 3) | ((b == 1) & (neighbors == 2))).astype(int)

board = np.zeros((10, 10), dtype=int)
board[4, 5:8] = 1
print(life_step(board))
```

This implementation wraps around at the edges.

## 100. Bootstrap 95% confidence interval for the mean
